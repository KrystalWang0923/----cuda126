```python

def is_blank_or_black(img_path, blank_thresh=240, black_thresh=15, var_thresh=10):
    try:
        img = Image.open(img_path).convert("L") #转灰度
        arr = np.array(img)#转换为数组
        mean = arr.mean()#计算平均值
        var = arr.var()#计算方差
        if mean > blank_thresh and var < var_thresh:
            return True#空白
        if mean < black_thresh and var < var_thresh:
            return True#黑板
        if var < var_thresh:
            return True#无用
        else:
            return None#文件有用
    except Exception as e:
        return True#文件损坏
```

```python
OK_list = []
NG_list = []
```

```python

for station_folder in os.listdir(root_path): #遍历各个工位
    station_folder_path = os.path.join(root_path, station_folder)
    if not os.path.isdir(station_folder_path):#判断是否为文件夹
        continue#不是文件夹就跳过
    for station_name in os.listdir(station_folder_path):#遍历各个工位中的文件夹（station1）
        station_path = os.path.join(station_folder_path, station_name)
        if not os.path.isdir(station_path):
            continue
        for date_name in os.listdir(station_path):#遍历每个日期下的文件
            date_path = os.path.join(station_path, date_name)
            if not os.path.isdir(date_path):
                continue
            for camera_name in os.listdir(date_path):#遍历每个相机下的文件
                camera_path = os.path.join(date_path, camera_name)
                if not os.path.isdir(camera_path):
                    continue
                for category in os.listdir(camera_path):#遍历OK和NG的图片
                    category_path = os.path.join(camera_path, category)
                    if not os.path.isdir(category_path):
                        continue
                    if category == "OK":
                        for dirpath,dirnames, filenames in os.walk(category_path):
                            for file in filenames:
                                if file.lower().endswith((".jpg", ".png", ".jpeg")):#判断格式
                                    OK_list.append(os.path.join(dirpath, file))
                    elif category == "NG":
                        for dirpath,dirnames, filenames in os.walk(category_path):
                            for file in filenames:
                                if file.lower().endswith((".jpg", ".png", ".jpeg")):
                                    NG_list.append(os.path.join(dirpath, file))
```
